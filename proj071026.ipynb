```python
# Tópico 11 - Atividade Prática
# Simulador de Gerenciamento de Memória

def fifo(paginas, quantidade_frames):
    memoria = []
    faltas = 0
    hits = 0

    for pagina in paginas:
        if pagina in memoria:
            hits += 1
            resultado = "HIT"
        else:
            faltas += 1
            resultado = "PAGE FAULT"

            if len(memoria) >= quantidade_frames:
                memoria.pop(0)
            memoria.append(pagina)

        print(f"Página: {pagina:<2} | Memória: {str(memoria):<20} | {resultado}")

    return faltas, hits


def lru(paginas, quantidade_frames):
    memoria = []
    faltas = 0
    hits = 0

    for pagina in paginas:
        if pagina in memoria:
            hits += 1
            resultado = "HIT"
            # Atualiza a página mais recentemente utilizada
            memoria.remove(pagina)
            memoria.append(pagina)
        else:
            faltas += 1
            resultado = "PAGE FAULT"

            if len(memoria) >= quantidade_frames:
                memoria.pop(0)
            memoria.append(pagina)

        print(f"Página: {pagina:<2} | Memória: {str(memoria):<20} | {resultado}")

    return faltas, hits


entrada = input("Digite as páginas separadas por espaço: ")
paginas_usuario = [int(valor) for valor in entrada.split()]

quantidade_frames = int(input("Digite a quantidade de frames: "))

if quantidade_frames <= 0 or len(paginas_usuario) == 0:
    print("Informe páginas e uma quantidade válida de frames.")
else:
    total = len(paginas_usuario)

    print("\nSequência:", paginas_usuario)
    print("Frames:", quantidade_frames)

    print("\n--- ALGORITMO FIFO ---")
    faltas_fifo, hits_fifo = fifo(paginas_usuario, quantidade_frames)

    taxa_faltas_fifo = (faltas_fifo / total) * 100
    taxa_hits_fifo = (hits_fifo / total) * 100

    print("\n--- RESULTADO FIFO ---")
    print("Page Faults:", faltas_fifo)
    print("Hits:", hits_fifo)
    print(f"Taxa de Page Faults: {taxa_faltas_fifo:.2f}%")
    print(f"Taxa de Hits: {taxa_hits_fifo:.2f}%")

    print("\n--- ALGORITMO LRU ---")
    faltas_lru, hits_lru = lru(paginas_usuario, quantidade_frames)

    taxa_faltas_lru = (faltas_lru / total) * 100
    taxa_hits_lru = (hits_lru / total) * 100

    print("\n--- RESULTADO LRU ---")
    print("Page Faults:", faltas_lru)
    print("Hits:", hits_lru)
    print(f"Taxa de Page Faults: {taxa_faltas_lru:.2f}%")
    print(f"Taxa de Hits: {taxa_hits_lru:.2f}%")

    print("\n--- COMPARAÇÃO DOS ALGORITMOS ---")
    print(f"FIFO: {faltas_fifo} Page Faults e {hits_fifo} Hits")
    print(f"LRU: {faltas_lru} Page Faults e {hits_lru} Hits")

    if faltas_fifo < faltas_lru:
        print("O algoritmo FIFO teve melhor desempenho.")
    elif faltas_lru < faltas_fifo:
        print("O algoritmo LRU teve melhor desempenho.")
    else:
        print("Os dois algoritmos tiveram o mesmo desempenho.")
```

# Tópico 13 -  Questões para análise

1. Qual é a diferença entre um HIT e um PAGE FAULT?

Resposta: HIT ocorre quando a página solicitada já está carregada na memória RAM. PAGE FAULT ocorre quando a página solicitada não está na memória, sendo necessário carregá-la.

2. Qual é a principal diferença entre FIFO e LRU?

Resposta: O FIFO remove a página que entrou primeiro na memória, independentemente de quantas vezes foi utilizada. Já o LRU remove a página que está há mais tempo sem ser acessada.

3. Um page fault significa necessariamente que existe um erro no programa? Explique.

Resposta: Não. O page fault é um acontecimento normal no gerenciamento de memória, pois indica que a página solicitada não está carregada na RAM. O sistema operacional precisa localizá-la e carregá-la, podendo substituir outra página caso a memória esteja cheia.

4. Aumentar a quantidade de frames sempre reduz o número de page faults no FIFO? Utilize o experimento da Anomalia de Belady para justificar.

Resposta: Não. Conforme demonstrado na Anomalia de Belady, aumentar a quantidade de frames pode provocar mais page faults no FIFO. No experimento da aula, utilizando 3 frames ocorreram 9 page faults, enquanto com 4 frames ocorreram 10. Portanto, aumentar a memória nem sempre melhora o desempenho desse algoritmo.

5. Em quais situações FIFO e LRU podem produzir resultados diferentes?

Resposta: Quando determinadas páginas são acessadas novamente, pois o FIFO considera apenas a ordem de entrada, enquanto o LRU considera o último acesso realizado. Assim, os algoritmos podem escolher páginas diferentes para substituir, produzindo quantidades distintas de hits e page faults.

6. Por que o algoritmo Ótimo normalmente apresenta um bom resultado?

Resposta: Porque ele considera os próximos acessos às páginas e remove aquela que será utilizada mais distante no futuro ou que não será mais utilizada. Dessa forma, consegue minimizar a quantidade de page faults para uma determinada sequência.

7. Por que um sistema operacional real não consegue utilizar perfeitamente o algoritmo Ótimo?

Resposta: Porque seria necessário conhecer antecipadamente todos os próximos acessos à memória. Como o sistema operacional não consegue prever exatamente quais páginas serão solicitadas no futuro, o algoritmo Ótimo é utilizado principalmente como referência teórica para comparar os demais algoritmos.
